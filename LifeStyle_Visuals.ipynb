```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Study = pd.read_csv(r"C:\Users\smrut\Desktop\MyWork\Matplotlib\Matplotlib_Day1_Day2_Datasets\Day1_MyStudyTracker.csv")
print(Study)
```

```python
plt.bar(Study["Day"],Study["Study Hours"],color = "purple", label = "Hours of Study")
plt.title("My Study Hours")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
daily_time = pd.read_csv(r"C:\Users\smrut\Desktop\MyWork\Matplotlib\Matplotlib_Day1_Day2_Datasets\Day2_DailyTime.csv")
print(daily_time)
```

```python
plt.pie(daily_time["Hours"],labels = daily_time["Activity"],autopct="%1.1f%%",colors = ["cyan","yellow","pink","purple","green","orange"])
plt.title("Regular Activity")
plt.legend( loc='upper right', bbox_to_anchor=(1.5, 1))
plt.tight_layout()
plt.show()
```

```python
fitness = pd.read_csv(r"C:\Users\smrut\Desktop\MyWork\Matplotlib\Matplotlib_Day1_Day2_Datasets\Day2_Fitness.csv")
print(fitness)
```

```python
plt.scatter(fitness["Sleep Hours"],fitness["Day"],marker = "^",color="blue",label="Sleep")
plt.scatter(fitness["Energy Level (1-10)"],fitness["Day"],marker = "*",color="purple",label="Energy")
plt.title("Comparison")
plt.xlabel("X-Axis")
plt.ylabel("Days")
plt.legend()
plt.tight_layout()
plt.show()
```
