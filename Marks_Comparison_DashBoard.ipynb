### Topic - Subplots & Layouts 
### "Marks Comparison Dashboard"

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\smrut\Desktop\MyWork\Matplotlib\Day3_MarksComparison.csv")
print(df)
```

```python
plt.bar(df["Subject"],df["Student A"],color = "y",width=0.5,label="Marks")
plt.title("Marks of Student A")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=60,ha="right")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.bar(df["Subject"],df["Student B"],color = "c",width=0.5,label="Marks")
plt.title("Marks of Student B")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=60,ha="right")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.bar(df["Subject"],df["Student C"],color = "m",width=0.5,label="Marks")
plt.title("Marks of Student C")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=60,ha="right")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.plot(df["Subject"],df["Student A"],marker='o',label = "Student A", color = "teal", linewidth=2)
plt.plot(df["Subject"],df["Student B"],marker='s',label = "Student B", color = "orange", linestyle = '--')
plt.plot(df["Subject"],df["Student C"],marker='^',label = "Student c", color = "green", linestyle = ':')

plt.title("Marks comaprison Dashboard")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=60,ha="right")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
df["avg_marks"] = df[["Student A","Student B","Student C"]].mean(axis = 1)
print(df[["Subject","avg_marks"]])
```

```python
plt.plot(df["Subject"],df["avg_marks"],color = "m",marker = "o",label = "Average Marks")
plt.title("Average marks of student in each subject")
plt.xlabel("Subject")
plt.ylabel("Average")
plt.xticks(rotation = 60 , ha = "right")
plt.legend()
plt.tight_layout()
plt.grid()
plt.show()
```
