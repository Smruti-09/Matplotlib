# MY STUDY TRACKER

```python
import matplotlib.pyplot as plt
```

```python
days = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
hours = [2.5,3.0,1.5,4.0,3.5,5.0,2.0]
plt.plot(days,hours,color = "cyan",marker="o",linewidth=2,linestyle=":",label="Daily Study Tracker")
plt.title("My Study Tracker")
plt.xlabel("Days")
plt.ylabel("Hours")
plt.legend()
plt.tight_layout()
plt.show()
```
